```python
import pandas as pd

PATH_TO_DATA = "hf://datasets/AbhayBhan/SalesData/s1.csv"

df = pd.read_csv(PATH_TO_DATA)
```

```python

# total revenue by product category
sales_by_product = df.groupby('Product line')['gross income'].sum().reset_index()
revenue = sales_by_product.sort_values(by='gross income', ascending=False).reset_index(drop=True)
print(revenue)
```

```python
# top 10 dates (in terms of revenue as the data doesn't contain customer name)

sales_by_branch = df.groupby('Date')['gross income'].sum().reset_index()

revenue_by_branch = sales_by_branch.sort_values(by='gross income', ascending=False).reset_index(drop=True)

print(revenue_by_branch[:10])
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df["Date"] = pd.to_datetime(df["Date"])
df["Month"] = df["Date"].dt.to_period("M")
monthly_sales = df.groupby("Month")["Total"].sum().reset_index()
monthly_sales["Month"] = monthly_sales["Month"].astype(str)
plt.figure(figsize=(8, 5))

plt.plot(
    monthly_sales["Month"],
    monthly_sales["Total"],
    marker="o",
    color="darkblue",
    linewidth=2,
)


plt.title("Total Monthly Sales Revenue", fontsize=14, fontweight="bold")
plt.xlabel("Month", fontsize=12)
plt.ylabel("Total Revenue ($)", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
# let's try to visualize weekly sales as monthly sales is very few
df["Week"] = df["Date"].dt.to_period("W")
monthly_sales = df.groupby("Week")["Total"].sum().reset_index()
monthly_sales["Week"] = monthly_sales["Week"].astype(str)
plt.figure(figsize=(15, 8))

plt.plot(
    monthly_sales["Week"],
    monthly_sales["Total"],
    marker="o",
    color="red",
    linewidth=2,
)


plt.title("Total Weekly Sales Revenue", fontsize=20, fontweight="bold")
plt.xlabel("Week", fontsize=16)
plt.xticks(rotation=45)
plt.ylabel("Total Revenue ($)", fontsize=16)
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
sales_by_product = df.groupby('Product line')['gross income'].sum().reset_index()
plt.figure(figsize=(8, 5))
plt.plot(
    sales_by_product['Product line'],
    sales_by_product['gross income'],
    marker="o",
    color="green",
    linewidth=2,
)

plt.title("Total Sales per Product", fontsize=14, fontweight="bold")
plt.xlabel("Month", fontsize=12)
plt.xticks(rotation=45)
plt.ylabel("Total Revenue ($)", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.tight_layout()
plt.show()
```
