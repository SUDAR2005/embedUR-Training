```python
import pandas as pd
import numpy as np
# read from huggingface
dataset_path = "hf://datasets/AbhayBhan/SalesData/s1.csv"
df = pd.read_csv(dataset_path)
```

```python
# fallback when offline or the data becomes unavailable
offline_path = "./sales_data.csv"
if df.empty:
    df = pd.read_csv(offline_path)
    print("Data is loaded from the offline copy")
else:
    print("Data is loaded from the online copy")
```

```python
# check for missing values (handled missing value using offline copy)
print(df.isnull().sum())
```

```python
# no missing value is found in online copy so using offline copy to demo handling missing value
# inducing missing value in 2 numerical and 3 categorical columns in offline copy

df_offline = pd.read_csv(offline_path)
print("The columns with missing values are")

column = df_offline.isnull().sum()
col_name = column[column > 0].index

summary_df = pd.DataFrame({
    'Missing Values': column[column > 0].values,
    'Data Type': df_offline.dtypes[column[column > 0].index]
}, index=col_name)

print(summary_df)
```

```python

# split based on Data Type
numerical_missing = summary_df[summary_df['Data Type'] != 'object'].index.tolist()
categorical_missing = summary_df[summary_df['Data Type'] == 'object'].index.tolist()

print("Numerical columns with missing values:", numerical_missing)
print("Categorical columns with missing values:", categorical_missing)
df_cleaned = df_offline.copy()

for col in numerical_missing:
    df_cleaned[col] = df_offline[col].fillna(df_offline[col].mean())

for col in categorical_missing:
    df_cleaned[col] = df_offline[col].fillna(df_offline[col].ffill())
```

```python
print(df_cleaned.isnull().sum())
df_cleaned.to_csv("./cleaned_sales_data.csv", index=False)
```
